```python
# Trying to get the wind with corrected sigma
# 1) Get sigma - thermal noise
from posada.Handlers.sentinel_1 import change_thermal_noise
import numpy as np
import matplotlib
import matplotlib.pyplot as plt



def get_values_thermalnoise(url, sigma0, inc):
# From  Sentinel_1(BaseSAR)._get_values_nice_image_hh_vv(self, slices=None, **kwargs) in posada/Handlers/sentinel-1.py
        from scipy.interpolate import interp1d
        from posada.Handlers.resources import sentinel_koeffs as sk
        from posada.Handlers.sentinel_1 import change_thermal_noise

        tnoise_new, flatline = change_thermal_noise(sk.tnoise, sk.incvec)
        fnoise_new = interp1d(sk.incvec, tnoise_new,
                              kind='slinear')
        
        sigma0 = np.ma.array(sigma0, mask=np.zeros_like(sigma0))
        sigma0_corr = sigma0 - fnoise_new(inc)

        return tnoise_new, sk.incvec, sigma0_corr, sigma0

inc = sar.get_values('incidence')
tnoise_n, incvec,  sigma0_corr, sigma0 = get_values_thermalnoise(pn+fn, s0, inc)  


plt.pcolormesh(np.log10(sigma0), cmap='jet',)
plt.axis('image')
plt.colorbar()
plt.show()



plt.pcolormesh(sigma0 - sigma0_corr)
plt.colorbar()
plt.show()
```

```python
from cmod.cmod_gpu import rcs2windOpenCl


PR = basesar.PR_Mouche(inc, wdir_interim)
try:
    from cmod.cmod_gpu import rcs2windOpenCl
    windSAR = rcs2windOpenCl(sar=sigma0 * PR,
                             windir=wdir_interim,
                             theta=inc)
except Exception:
    from cmod.cmod_vect import rcs2windPar
    windSAR = rcs2windPar(sar=sigma0 * PR, cmdv=5,
                          windir=wdir_interim,
                          theta=inc,
                          nprocs=4)

    
    
try:
    from cmod.cmod_gpu import rcs2windOpenCl
    windSAR_corr = rcs2windOpenCl(sar=sigma0_corr * PR,
                                  windir=wdir_interim,
                                  theta=inc)
except Exception:
    print ('exception')
    from cmod.cmod_vect import rcs2windPar
    windSAR_corr = rcs2windPar(sar=sigma0_corr * PR, cmdv=5,
                               windir=wdir_interim,
                               theta=inc,
                               nprocs=4)
    
```

```python
importlib.reload(wnd)
# Plotting the sections along the mean wind direction

distance = 10e4
xs = [250, 270, 360]
ys = [220, 260, 300]

window_size = 2    # cut of the highest frequencies corresponding to x points 

window_size_km = window_size*resolution/1000 #optional - need for plotting 


#try to extract multiple sections at different places


ccc = ['black', 'blue', 'magenta', 'red']
section_coords = {'x0':[], 'x1':[], 'y0':[], 'y1':[]}
distance_vector = {}
data_section = {}
polynoms = {}

n = 0
for x0, y0 in zip(xs, ys):
#         print (x0, y0)
        data_section[n], distance_vector[n], section_coords_temp = wnd.section_extract(wind_speed, x0, y0, \
                                                                                   np.mean(wdir_interim.squeeze()),\
                                                                                   distance, resolution)
        section_coords['x0'].append(section_coords_temp[0])
        section_coords['x1'].append(section_coords_temp[1])
        section_coords['y0'].append(section_coords_temp[2])
        section_coords['y1'].append(section_coords_temp[3])
        n+=1

        
# Mask out "high winds" ~ sea ice
for n in np.arange(len(data_section)):
    data_section[n] = ma.masked_greater(data_section[n], 18)



# Plotplotplot

fig_sections = plt.figure(figsize=(18, 30))
plt.subplot(3, 1, 1)
wnd.wind_plot(fn, wind_speed, wdir_interim, lat, lon, every_x = 50, case = 'ERA INTERIM', \
              section_coords=section_coords, sect_cols=ccc)
# wnd.wind_plot(fn, wind_speed, wdir_interim, every_x = 50, case = 'ERA INTERIM',\
#                                    lat=[], lon =[], section_coords=section_coords, sect_cols=ccc)


plt.axis('image')


plt.subplot(3, 1, 2)
for n in np.arange(len(section_coords['x0'])):
    print(n)   
    distance_km = distance_vector[n]/1000 
    plt.plot(distance_km, data_section[n], color = ccc[n])

plt.title("Wind speed along sections")   
plt.xlabel('distance, km')
plt.ylabel('wind speed, m/s')



plt.subplot(3, 1, 3)
for n in np.arange(len(section_coords['x0'])):
    N  = 5    # Filter order
    Wn = 1/window_size   #cut-off _spatial_ frequency (cut the highest frequencies corresponding to n_points)
    B, A = sg.butter(N, Wn, output='ba')

    # Second, apply the filter
    data_f = sg.filtfilt(B, A, data_section[n])
    plt.plot(distance_km, data_f, label = 'section '+str(n), color = ccc[n])

    deg = 3
    z = np.polyfit(distance_km, data_section[n], deg)
    polynoms[n] = np.poly1d(z) #polynomial structure (special datatype in python)
#     plt.plot(distance_km, p(distance_km), label = str(z[0])+'x +'+str(z[1]))    
    plt.plot(distance_km, polynoms[n](distance_km), label = str(polynoms[n]), linewidth = 3)    

    
plt.legend()
plt.title("Butterworth filter, window size "+str(window_size_km)+' km')
plt.xlabel('distance, km')
plt.ylabel('wind speed, m/s')

plt.tight_layout()
plt.show()

if area_x!=[] and area_y!=[]:
    figname = '/home/tarasenko/Figures/tests_wind/'+fn[:-4]+'-'+str(area_x[0])\
    +'-'+str(area_x[1])+'-'+str(area_y[0])+'-'+str(area_y[1])+'_wind_sections.png'
else:
    figname = '/home/tarasenko/Figures/tests_wind/'+fn[:-4]+'_wind_sections.png'
    
fig_sections.savefig(figname, dpi = 300)
```
